```python
import os
# iterate over all the directories where the test
for class_folder in os.listdir(test_folder):
  class_folder_path = os.path.join(test_folder, class_folder)
  if os.path.isdir(class_folder_path):
    ground_truth_class = class_folder
    print(f"Processing Class for (Ground truth) {ground_truth_class}")
    image_files = [os.path.join(class_folder_path, img) for img in os.listdir(class_folder_path)]
    for img_path in image_files:
      results = model.predict(img_path, conf=0.25)
      for r in results:
        predicted_class_index = r.probs.top1 # get the predicted class
        print(f"class_name:{r.names[predicted_class_index]}")
        print(f"conf:{r.probs.top1conf}")



```
